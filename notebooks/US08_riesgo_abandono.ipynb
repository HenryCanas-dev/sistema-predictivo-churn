# US-08: Calcular riesgo de abandono
- Rol: Como analista de negocio.   
- Objetivo: Consultar la probabilidad de abandono de un cliente para tomar acciones preventivas de retención
- Criterios de aceptación:
Mostrar porcentaje de riesgo.   
Clasificar el riesgo en Bajo, Medio o Alto. 

```python
import pandas as pd
import numpy as np

def calcular_riesgo_cliente(probabilidad_churn: float) -> dict:
    """
    Toma la probabilidad predicha por el modelo de ML (entre 0 y 1)
    y retorna el porcentaje formateado y la categoría de riesgo (US-08).
    """
    # 1. Convertir la probabilidad a porcentaje
    porcentaje_riesgo = round(probabilidad_churn * 100, 2)
    
    # 2. Clasificar el riesgo (Criterio de aceptación US-08)
    if probabilidad_churn >= 0.70:
        categoria = "Alto"
        accion_sugerida = "Ofrecer descuento inmediato y llamada del equipo de retención."
    elif probabilidad_churn >= 0.40:
        categoria = "Medio"
        accion_sugerida = "Enviar encuesta de satisfacción y promoción ligera."
    else:
        categoria = "Bajo"
        accion_sugerida = "Sin acción requerida. Cliente fidelizado."
        
    return {
        "porcentaje_riesgo": f"{porcentaje_riesgo}%",
        "probabilidad_raw": probabilidad_churn,
        "categoria_riesgo": categoria,
        "accion_sugerida": accion_sugerida
    }

#PRUEBAS CON CLIENTES DE EJEMPLO

# Ejemplo 1: Cliente con alta probabilidad de irse
cliente_riesgoso = calcular_riesgo_cliente(0.85)
print("Cliente 1:", cliente_riesgoso)

# Ejemplo 2: Cliente con riesgo moderado
cliente_medio = calcular_riesgo_cliente(0.52)
print("Cliente 2:", cliente_medio)

# Ejemplo 3: Cliente leal
cliente_seguro = calcular_riesgo_cliente(0.15)
print("Cliente 3:", cliente_seguro)
```
